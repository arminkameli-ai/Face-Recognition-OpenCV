```python
import face_recognition as fr
import cv2 
import numpy as np
```

```python
image = fr.load_image_file("biden.jpg")
sabt_sorat = fr.face_encodings(image)[0]

image_farzan = fr.load_image_file("obama.jpg")

sabt_sorat2 = fr.face_encodings(image_farzan)[0]
```

```python
cap = cv2.VideoCapture(0)
```

```python
while True:

    ret, img = cap.read()

    if not ret:
        break

    img2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    locate = fr.face_locations(img2)

    if len(locate) > 0:

        img = cv2.rectangle(
            img,
            (locate[0][3], locate[0][0]),
            (locate[0][1], locate[0][2]),
            (0, 255, 0),
            3,
        )

        encode = fr.face_encodings(img2)

        result = fr.compare_faces(
            [sabt_sorat, sabt_sorat2],
            encode[0]
        )

        if result[0]:
            cv2.putText(
                img,
                "biden",
                (locate[0][3], locate[0][0]-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0,255,0),
                2
            )

        elif result[1]:
            cv2.putText(
                img,
                "obama",
                (locate[0][3], locate[0][0]-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0,255,0),
                2
            )

        else:
            cv2.putText(
                img,
                "Unknown",
                (locate[0][3], locate[0][0]-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0,0,255),
                2
            )

    cv2.imshow("Face Recognition", img)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
